## 5. Operators
Operators do maths (`+ - * / // % **`), compare values (`== != > < >= <=`), combine yes/no answers (`and or not`) and update boxes (`+= -= *= /=`).

### Arithmetic
- `/`: division always gives a decimal: 2.0
- `//`: floor division: how many whole times 2 fits in 11 (5)
- `%`: remainder: what is left over (1)
- `**`: power: 2 x 2 x 2 = 8

```
print(10 + 5)
print(10 - 5)
print(10 * 5)
print(10 / 5)
print(11 // 2)
print(11 % 2)
print(2 ** 3)
```

### Comparisons
- `==`: is equal to (two signs; one = means "put in the box")
- `!=`: is not equal to
- `>=  <=`: greater/less than or equal to
- `True / False`: every comparison answers with a bool

```
x = 10
y = 20
print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

### and, or, not
- `and`: True only if BOTH sides are True
- `or`: True if AT LEAST ONE side is True
- `not`: flips True to False and False to True

```
age = 25
has_license = False
print(age >= 18 and has_license)
print(age >= 18 or has_license)
print(not has_license)
```

### Shortcuts for updating
- `x += 5`: take what is in x, add 5, put the result back in x
- `x /= 2`: division gives a decimal, so x becomes 19.5

```
x = 10
x += 5     # x = x + 5
print(x)
x -= 2     # x = x - 2
print(x)
x *= 3     # x = x * 3
print(x)
x /= 2     # x = x / 2
print(x)
```

## 6. Strings
A string is text. Each character has a position (**index**) starting at 0. Negative indexes count from the end. A **slice** `[start:stop]` cuts out a piece and stops *before* `stop`.

### Indexing
- `word[0]`: the first character (counting starts at 0)
- `word[-1]`: the last character

```
word = "Python"
print(word[0])
print(word[1])
print(word[-1])
```

### Slicing
- `word[0:3]`: positions 0, 1, 2 (stops before 3): "Pyt"
- `word[2:]`: from position 2 to the end
- `word[:4]`: from the start up to (not including) 4
- `word[::-1]`: step backwards: the word reversed

```
print(word[0:3])
print(word[2:])
print(word[:4])
print(word[::-1])
```

### String methods
- `.upper()`: a method: a function that belongs to a value, called with a dot. Gives a CAPITALS copy
- `.strip()`: removes spaces from both ends
- `.replace(a, b)`: swaps every a for b
- `.split()`: cuts the text into a list of words

```
text = "  machine learning  "
print(text.upper())
print(text.strip())
print(text.replace("machine", "deep"))
print(text.split())
```

### f-strings: fill in the blanks
- `f"..."`: the f before the quotes turns on fill-in-the-blanks
- `{name}`: replaced by whatever is in the name box

```
name = "John"
age = 25
print(f"My name is {name} and I am {age} years old.")
```

## 7. Data structures: lists, tuples, sets, dictionaries
A **list** is an ordered, changeable row of items. A **tuple** is a list that can't be changed. A **set** keeps only unique items. A **dictionary** stores **key: value** pairs.

### Lists
- `[ ... ]`: square brackets make a list; commas separate the items
- `.append(100)`: adds 100 to the end
- `.insert(1, 88)`: puts 88 at position 1, pushing the rest along
- `.remove(75)`: deletes the first 75 it finds

```
marks = [80, 90, 75, 85, 95]
print(marks[0])
print(marks[-1])
marks.append(100)
marks.insert(1, 88)
marks.remove(75)
print(marks)
```

### Useful list tools
- `len / sum / max / min`: count, add up, biggest, smallest
- `.sort()`: puts the list in order, smallest first
- `marks[0] = 82`: replace one item: lists are mutable (changeable)

```
print(len(marks), sum(marks), max(marks), min(marks))
marks.sort()
print(marks)
marks[0] = 82   # lists can be changed
print(marks)
data = ["John", 25, 1.75, True]   # mixed types are fine
```

### Tuples: a locked list
- `( ... )`: round brackets make a tuple
- `point[0] = 5`: not allowed: tuples are immutable, so Python raises an error

```
point = (10, 20)
print(point[0])
print(point[1])
point[0] = 5
```

### Sets: unique items only
- `{ ... }`: curly brackets with single values make a set
- `duplicates`: repeated values are kept only once

```
ids = {101, 102, 102, 103, 101}
print(ids)
```

### Set operations
- `union`: everything in a or b
- `intersection`: only what is in both
- `difference`: in a but not in b

```
a = {1, 2, 3}
b = {3, 4, 5}
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
```

### Dictionaries
- `"name": "John"`: a key (the label) and its value
- `student["name"]`: look up a value by its key
- `.get("marks")`: another way to look up
- `student["grade"] = "A"`: a new key adds a new pair; an existing key updates it

```
student = {"name": "John", "age": 25, "marks": 85}
print(student["name"])
print(student.get("marks"))
student["age"] = 26
student["grade"] = "A"
print(student)
```

### Keys, values and missing keys
- `.keys() / .values()`: all the labels / all the values
- `.get("city")`: a missing key gives None instead of an error

```
print(student.keys())
print(student.values())
print(student.get("city"))
```

## 8. Conditions: if, elif, else
Conditions let a program choose. Python checks each condition from the top and runs only the **first** block that is True. Indented lines (4 spaces) belong to the line above them.

### if / elif / else
- `if marks >= 80:`: a question, then a colon
- `    print(...)`: indented: only runs if the question above is True
- `elif`: "else if": checked only if everything above was False
- `else:`: runs if nothing above was True

```
marks = 75
if marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")
```

### Nested conditions
- `an if inside an if`: the inner question is only asked when the outer answer is yes
- `8 spaces`: two levels of indentation

```
age = 22
has_license = True
if age >= 18:
    if has_license:
        print("Allowed to drive")
    else:
        print("License required")
else:
    print("Too young")
```

### One-line if
- `A if condition else B`: gives A when the condition is True, otherwise B

```
marks = 75
result = "Pass" if marks >= 50 else "Fail"
print(result)
```

## 9. Loops: for, while, break, continue
A **loop** repeats a block of code. `for` goes through items one at a time. `while` repeats as long as a condition stays True.

### for and range
- `range(1, 6)`: the numbers 1, 2, 3, 4, 5: it stops BEFORE 6
- `number`: holds the next value each time around

```
for number in range(1, 6):
    print(number)
```

### Looping over a list and a dictionary
- `for mark in marks`: each item of the list, in order
- `.items()`: gives key and value pairs; we unpack them into two names

```
marks = [80, 90, 75]
for mark in marks:
    print(mark)

for key, value in student.items():
    print(key, value)
```

### while
- `while count <= 5:`: keep going as long as this is True
- `count += 1`: without this the condition never becomes False: an infinite loop

```
count = 1
while count <= 5:
    print(count)
    count += 1
```

### break: stop early
- `range(10)`: 0 to 9 (one number starts at 0)
- `break`: leave the loop immediately

```
for number in range(10):
    if number == 5:
        break
    print(number)
```

- `continue`: skip the rest of this round and go to the next one

```
for number in range(5):
    if number == 2:
        continue
    print(number)
```

```
for row in range(3):
    for seat in range(2):
        print(row, seat)
```

## 10. Functions
A **function** is a reusable machine: inputs go in (**parameters**), it does its work, and **return** hands a result back.

### Defining and calling
- `def greet():`: def = define: build a function called greet
- `greet()`: call it: brackets make it run

```
def greet():
    print("Hello!")

greet()
```

### Parameters and arguments
- `name`: a parameter: the box that receives the input
- `"John"`: an argument: the actual value we pass in

```
def greet(name):
    print(f"Hello {name}")

greet("John")
greet("Alice")
```

### return
- `return a + b`: hand the answer back to whoever called the function
- `result = add(10, 20)`: the returned 30 goes into result

```
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

### return vs print
- `print inside`: shows 30 on screen, but hands nothing back
- `answer`: so answer is None

```
def show_total(a, b):
    print(a + b)

answer = show_total(10, 20)
print(answer)
```

### Default and keyword arguments
- `name="friend"`: a default: used when no argument is given
- `age=25, name="John"`: keyword arguments: named, so order does not matter

```
def greet(name="friend"):
    print(f"Hello {name}")

greet()
greet("Alice")

def student_info(name, age):
    print(name, age)

student_info(age=25, name="John")
```

### *args and **kwargs
- `*numbers`: accepts any number of values, collected into a tuple
- `**details`: accepts any number of named values, collected into a dictionary

```
def calculate_total(*numbers):
    return sum(numbers)

print(calculate_total(10, 20, 30, 40))

def display_info(**details):
    for key, value in details.items():
        print(key, value)

display_info(name="John", age=25, city="London")
```

### Scope: where a variable lives
- `x`: global: made outside any function, visible everywhere
- `inside`: local: made inside the function, gone when it finishes
- `print(inside)`: error: inside does not exist outside example()

```
x = 10   # global

def example():
    inside = 20   # local
    print(x, inside)

example()
print(x)
print(inside)
```

### lambda: a one-line function
- `lambda x: x * x`: a tiny function with no name: takes x, gives back x * x

```
square = lambda x: x * x
print(square(5))
```

## 11. Comprehensions
A **comprehension** builds a list, dictionary or set in one line: *what to keep* `for` *each item* `in` *something*, optionally `if` *condition*.

### Loop vs comprehension
- `[number ** 2 for number in range(1, 6)]`: the same four lines of loop, in one line

```
squares = []
for number in range(1, 6):
    squares.append(number ** 2)
print(squares)

squares = [number ** 2 for number in range(1, 6)]
print(squares)
```

### With a condition, dict and set
- `if x % 2 == 0`: keep only numbers with remainder 0 when divided by 2 (even numbers)
- `{x: x ** 2 ...}`: key: value makes a dictionary
- `{x ** 2 ...}`: single values make a set, so duplicates disappear

```
even = [x for x in range(1, 11) if x % 2 == 0]
print(even)

squares = {x: x ** 2 for x in range(1, 6)}
print(squares)

unique = {x ** 2 for x in [1, 2, 2, 3, 3]}
print(unique)
```

## 12. Errors and exceptions
When something goes wrong, Python stops and shows an **error**. Read the **last line** first: it names the error and says why. `try` / `except` catch errors so the program can carry on.

### An error
- `ValueError`: the kind of error: a value that cannot be used this way
- `last line`: always read it first: it says what went wrong

```
number = int("hello")
```

### try / except
- `try:`: code that might fail
- `except ValueError:`: what to do if that particular error happens

```
try:
    number = int("hello")
except ValueError:
    print("Please enter a valid number.")
```

### else and finally
- `else:`: runs only if no error happened
- `finally:`: runs no matter what

```
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print(result)
finally:
    print("Program finished.")
```

### raise your own error
- `raise`: stop the program on purpose, with your own message

```
age = -5
if age < 0:
    raise ValueError("Age cannot be negative")
```

## 14. Files: text, CSV and JSON
Programs can save data to **files** and read it back later. `with open(...)` opens a file and closes it automatically. Modes: `r` read, `w` write (replaces), `a` append, `x` create new.

### Write and read a text file
- `open("students.txt", "w")`: open for writing (creates the file, or wipes it)
- `\n`: a new line character
- `with ... as file:`: the file is closed automatically at the end of the block
- `file.read()`: reads everything as one string

```
with open("students.txt", "w") as file:
    file.write("John\n")
    file.write("Alice\n")

with open("students.txt", "r") as file:
    content = file.read()
print(content)
```

### Append
- `"a"`: append: add to the end without wiping
- `open(name)`: with no mode, "r" (read) is used

```
with open("students.txt", "a") as file:
    file.write("David\n")

with open("students.txt") as file:
    print(file.read())
```

### CSV: a table as text
- `csv.writer(file)`: a helper that writes rows with commas between values
- `newline=""`: stops blank lines appearing between rows on some computers
- `writerow([...])`: one list = one row

```
import csv

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "marks"])
    writer.writerow(["John", 85])
    writer.writerow(["Alice", 90])

with open("students.csv") as file:
    print(file.read())
```

### JSON: save a dictionary
- `json.dump`: save a dictionary to a file (indent=4 makes it readable)
- `json.load`: read it back as a dictionary

```
import json

student = {"name": "John", "age": 25, "marks": 85}
with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

with open("student.json") as file:
    data = json.load(file)
print(data["name"])
```

## 15. Classes and objects (OOP)
A **class** is a blueprint; an **object** is a thing built from it. `__init__` runs when an object is created; `self` means *this object*. Data on an object are **attributes**; functions on it are **methods**.

### A class
- `class Student:`: the blueprint (class names start with a capital)
- `__init__`: the constructor: runs automatically when an object is made
- `self.name = name`: store the value on this particular object (an attribute)
- `def introduce(self)`: a method: a function that belongs to the object
- `Student("John", 25)`: build an object from the blueprint

```
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name}")

student1 = Student("John", 25)
student2 = Student("Alice", 22)
student1.introduce()
student2.introduce()
print(student2.age)
```

### Inheritance
- `class Dog(Animal):`: Dog inherits from Animal: it gets everything Animal has
- `dog.speak()`: works even though Dog never defines speak

```
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

dog = Dog()
dog.speak()
dog.bark()
```

## 16. Handy tools: map, filter, zip, enumerate, iterators, generators
Short tools you will see constantly in real-world code.

### map and filter
- `map(f, items)`: apply f to every item
- `filter(f, items)`: keep items where f says True
- `list( )`: turn the result into a list we can print

```
numbers = [1, 2, 3, 4, 5, 6]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
```

### zip and enumerate
- `zip(a, b)`: pairs items up like the teeth of a zip
- `enumerate(names)`: gives each item with its position

```
names = ["John", "Alice", "David"]
marks = [85, 90, 78]
print(list(zip(names, marks)))

for index, name in enumerate(names):
    print(index, name)
```

### Iterators
- `iter( )`: makes an iterator: hands out values one at a time
- `next( )`: ask for the next value

```
numbers = iter([10, 20, 30])
print(next(numbers))
print(next(numbers))
print(next(numbers))
```

### Generators
- `yield`: hand out one value, pause, and continue from here next time
- `why`: values are made only when needed, which saves memory on big data

```
def count_up():
    yield 1
    yield 2
    yield 3

for number in count_up():
    print(number)
```

## 17. Mini-project: student reports
Everything together: a list of dictionaries, functions, conditions, loops and f-strings.

### The data and the average
- `a list of dictionaries`: one dictionary per student, all in one list
- `sum / len`: total divided by how many = the average

```
students = [
    {"name": "John", "marks": [80, 90, 85]},
    {"name": "Alice", "marks": [95, 92, 88]},
    {"name": "David", "marks": [45, 55, 60]},
]

def calculate_average(marks):
    return sum(marks) / len(marks)

print(calculate_average([80, 90, 85]))
```

### Grades
- `return inside if`: the function stops at the first return it reaches

```
def calculate_grade(average):
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "Fail"

print(calculate_grade(85))
```

### The report
- `student["marks"]`: look up the list of marks in this student's dictionary
- `{average:.2f}`: show the number with 2 decimal places
- `for student in students`: one report per student

```
def display_report(student):
    average = calculate_average(student["marks"])
    grade = calculate_grade(average)
    print(f"{student['name']}: average {average:.2f}, grade {grade}")

for student in students:
    display_report(student)
```
