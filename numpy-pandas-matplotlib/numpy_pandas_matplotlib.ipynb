## 1. NumPy: maths on whole lists of numbers
NumPy gives Python the **array**: a grid of numbers where maths happens on every number at once. Almost every machine learning tool is built on it.

### List vs array
- `import numpy as np`: load NumPy and give it the short nickname np (everyone uses it)
- `marks * 2`: a plain list just repeats itself
- `np.array(marks)`: turn the list into a NumPy array
- `arr * 2,  arr + 5`: maths is applied to every number at once (vectorised)

```
import numpy as np

marks = [71, 57, 59, 64]
print(marks * 2)

arr = np.array(marks)
print(arr * 2)
print(arr + 5)
```

### Quick ways to make arrays
- `np.arange(0, 10, 2)`: start at 0, stop before 10, step 2
- `np.linspace(0, 1, 5)`: 5 evenly spaced numbers from 0 to 1, both ends included
- `np.zeros(3) / np.ones(3)`: arrays filled with 0s or 1s, handy as a starting point

```
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.zeros(3))
print(np.ones(3))
```

### 2D arrays: a table of numbers
- `[[...], [...], [...]]`: a list of rows: 3 students (rows) x 3 tests (columns)
- `.shape`: (rows, columns) = (3, 3)
- `.ndim`: number of dimensions: 2 for a table
- `.dtype`: the type of every number: int64 = whole numbers

```
scores = np.array([[71, 65, 80],
                   [57, 62, 49],
                   [59, 70, 66]])
print(scores)
print(scores.shape)
print(scores.ndim)
print(scores.dtype)
```

### Picking rows, columns and cells
- `scores[0]`: the first row (counting from 0)
- `scores[0, 2]`: row, then column
- `scores[:, 1]`: : means every row, so this is the whole column 1
- `scores[1:, :2]`: slices work in both directions at once

```
print(scores[0])        # row 0: John
print(scores[0, 2])     # row 0, column 2
print(scores[:, 1])     # every row, column 1
print(scores[1:, :2])   # rows 1 onwards, first 2 columns
```

### Totals and averages, by row or by column
- `.sum() / .mean() / .max()`: over every number in the table
- `axis=1`: go across each row: one answer per student
- `axis=0`: go down each column: one answer per test

```
print(scores.sum())
print(scores.mean())
print(scores.max())
print(scores.mean(axis=1))   # each student's average
print(scores.mean(axis=0))   # each test's average
```

### Broadcasting
- `scores + bonus`: the 3-number bonus is stretched over every row automatically: broadcasting
- `scores / 100`: one number is stretched over the whole table

```
bonus = np.array([5, 0, 10])   # a bonus for each test
print(scores + bonus)
print(scores / 100)
```

### True/False masks
- `scores > 60`: asks the question for every number: a table of True/False
- `scores[scores > 60]`: keep only the numbers where the answer is True
- `(...).sum()`: True counts as 1, so this counts the matches

```
print(scores > 60)
print(scores[scores > 60])
print((scores > 60).sum())
```

### Random numbers and reshape
- `default_rng(42)`: a random number generator; 42 is the seed, so you get the same numbers every run
- `integers(1, 7, size=5)`: 5 whole numbers from 1 up to (not including) 7
- `.reshape(2, 3)`: same 6 numbers, rearranged into 2 rows x 3 columns

```
rng = np.random.default_rng(42)
print(rng.integers(1, 7, size=5))   # 5 dice rolls
print(np.arange(6).reshape(2, 3))
```

## 2. pandas: tables with names
pandas gives Python the **DataFrame**: a table with named columns, like a spreadsheet. A single column is a **Series**.

### A Series: a labelled column
- `import pandas as pd`: load pandas with its usual nickname pd
- `pd.Series(values, index=...)`: a column of values, each with a label
- `marks["Alice"]`: look a value up by its label

```
import pandas as pd

marks = pd.Series([71, 57, 59], index=["John", "Alice", "David"])
print(marks)
print(marks["Alice"])
```

### Loading a CSV file
- `url = (...)`: the web address of our students file (a CSV: a table saved as text); strings side by side inside ( ) join into one
- `pd.read_csv(url)`: read the file into a DataFrame called df
- `df.head()`: show the first 5 rows

```
url = ("https://raw.githubusercontent.com/Uttam-Nepal/"
       "machine-learning-explained/main/"
       "numpy-pandas-matplotlib/students.csv")
df = pd.read_csv(url)
df.head()
```

### First look: shape, columns, counts
- `.shape`: (rows, columns): 20 students, 6 columns
- `.columns.tolist()`: the column names, as a list
- `df.count()`: how many values are filled in, in each column: sleep and attendance have one gap each

```
print(df.shape)
print(df.columns.tolist())
print(df.count())
```

### Summary numbers
- `count`: how many values are filled in
- `mean / min / max`: average, smallest, biggest
- `50%`: the middle value when sorted (25% and 75% are the quarter points)
- `std`: standard deviation: a measure of how spread out the values are

```
df.describe()
```

### Selecting columns
- `df["marks"]`: one column: a Series
- `df[["name", "marks"]]`: a list of columns: a smaller DataFrame (note the double brackets)

```
print(df["marks"].head(3))
print(df[["name", "marks"]].head(3))
```

### Selecting rows: loc and iloc
- `df.loc[row, column]`: by label (here the row labels are 0, 1, 2...)
- `df.iloc[-1]`: by position: -1 is the last row
- `loc[0:2, ...]`: with loc, the end of a range IS included: rows 0, 1 and 2

```
print(df.loc[0, "name"], df.loc[0, "marks"])
print(df.iloc[-1]["name"])
df.loc[0:2, ["name", "hours"]]
```

### Filtering rows
- `df["hours"] >= 13`: a True/False mask, one per row
- `df[mask]`: keep only the rows where the mask is True

```
df[df["hours"] >= 13]
```

### Two conditions
- `&`: and: both conditions must be True (use | for or)
- `( ) around each`: required, so Python checks each condition first

```
df[(df["hours"] >= 10) & (df["sleep"] >= 7)]
```

### Sorting
- `sort_values("marks")`: order the rows by marks
- `ascending=False`: biggest first

```
df.sort_values("marks", ascending=False).head(3)
```

### New columns
- `df["percent"] = ...`: a new column, calculated for every row at once
- `np.where(condition, a, b)`: an if/else for a whole column: "high" where True, "low" where False

```
df["percent"] = df["marks"] / 100
df["study_level"] = np.where(df["hours"] >= 10, "high", "low")
df.head(3)
```

### Finding missing values
- `.isna()`: True wherever a value is missing (NaN = not a number)
- `.sum()`: count the missing values in each column

```
print(df.isna().sum())
```

### Fixing missing values
- `fillna(df["sleep"].mean())`: fill the gap with the average sleep
- `dropna()`: drop any row that still has a gap (Grace's attendance)
- `.sum().sum()`: total missing values left: 0

```
df["sleep"] = df["sleep"].fillna(df["sleep"].mean())
df = df.dropna()
print(df.isna().sum().sum())
print(df.shape)
```

### Group and compare
- `groupby("passed")`: split the rows into groups: yes and no
- `[[...]].mean()`: average of each column, inside each group
- `.round(1)`: one decimal place

```
df.groupby("passed")[["hours", "sleep", "marks"]].mean().round(1)
```

### Counting categories
- `.value_counts()`: how many rows have each value

```
df["study_level"].value_counts()
```

### Saving the clean table
- `to_csv`: save the DataFrame as a CSV file
- `index=False`: don't save the row numbers as an extra column

*In Colab the file appears in the Files panel on the left.*

```
df.to_csv("students_clean.csv", index=False)
print(pd.read_csv("students_clean.csv").shape)
```

## 3. Matplotlib: pictures of data
Matplotlib draws charts. The pattern is always the same: draw, add a title and axis labels, show.

### Line chart: change over time
- `import matplotlib.pyplot as plt`: the charting module, nickname plt
- `plt.plot(x, y)`: a line through the points; marker='o' draws a dot at each point
- `title / xlabel / ylabel`: always label your chart
- `plt.show()`: draw it

```
import matplotlib.pyplot as plt

weeks = [1, 2, 3, 4, 5, 6]
hours_week = [5, 7, 6, 9, 11, 12]
plt.plot(weeks, hours_week, marker="o")
plt.title("John's study hours")
plt.xlabel("week")
plt.ylabel("hours")
plt.show()
```

### Bar chart: compare things
- `top = ...head(5)`: the 5 best students
- `plt.bar(labels, heights)`: one bar per student

```
top = df.sort_values("marks", ascending=False).head(5)
plt.bar(top["name"], top["marks"])
plt.title("Top 5 marks")
plt.ylabel("marks")
plt.show()
```

### Histogram: how values are spread
- `plt.hist(values, bins=6)`: cut the range of marks into 6 buckets and count students in each
- `edgecolor="black"`: outline each bar so the buckets are easy to see

```
plt.hist(df["marks"], bins=6, edgecolor="black")
plt.title("How marks are spread")
plt.xlabel("marks")
plt.ylabel("number of students")
plt.show()
```

### Scatter plot: is there a relationship?
- `plt.scatter(x, y)`: one dot per student: hours across, marks up
- `the pattern`: dots rising to the right = more hours tends to mean higher marks

```
plt.scatter(df["hours"], df["marks"])
plt.title("Study hours vs marks")
plt.xlabel("hours studied per week")
plt.ylabel("marks")
plt.show()
```

### Two charts side by side
- `plt.subplots(1, 2)`: a figure with 1 row and 2 chart spaces (axes)
- `figsize=(10, 4)`: width and height in inches
- `axes[0] / axes[1]`: draw on the left or right chart; set_title instead of title

```
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].scatter(df["sleep"], df["marks"])
axes[0].set_title("Sleep vs marks")
axes[1].scatter(df["attendance"], df["marks"])
axes[1].set_title("Attendance vs marks")
plt.show()
```

## 4. Mini project: what goes with passing?
Everything together: filtering, a scatter plot with a legend, and group averages.

### Who passed?
- `passed / failed`: two filtered tables, one per group
- `color=..., label=...`: each group gets its own colour and a name for the legend
- `plt.axhline(50)`: a horizontal dashed line at the pass mark
- `plt.legend()`: a key explaining the colours, so the chart doesn't rely on colour alone

```
passed = df[df["passed"] == "yes"]
failed = df[df["passed"] == "no"]
plt.scatter(passed["hours"], passed["marks"], color="green", label="passed")
plt.scatter(failed["hours"], failed["marks"], color="red", label="did not pass")
plt.axhline(50, linestyle="--", color="gray", label="pass mark")
plt.legend()
plt.title("Who passed?")
plt.xlabel("hours studied per week")
plt.ylabel("marks")
plt.show()
```

### Back it up with numbers
- `passed["hours"].mean()`: average study hours of the students who passed
- `round(..., 1)`: one decimal place

```
print("Average hours, passed:", round(passed["hours"].mean(), 1))
print("Average hours, did not pass:", round(failed["hours"].mean(), 1))
```
