# What is Machine Learning? Your first model
We teach a computer to tell **apples** from **oranges**, two ways:

1. **We** write the rule (normal programming)
2. The **computer learns** the rule from examples (machine learning)

Run each cell with **Shift + Enter**.

## Step 1: the data
8 fruits. We measured **weight** (grams) and **smoothness** (1 = very bumpy, 10 = very smooth), and we know each answer.

```
import pandas as pd

fruits = pd.DataFrame({
    "weight":     [150, 130, 175, 140, 170, 180, 145, 190],
    "smoothness": [  8,   9,   7,   9,   3,   2,   4,   3],
    "label":      ["apple"]*4 + ["orange"]*4,
})
fruits
```

## Step 2: WE write the rule
Our guess: heavier than 160 g means orange.

```
def my_rule(weight):
    if weight > 160:
        return "orange"
    return "apple"

guesses = [my_rule(w) for w in fruits.weight]
correct = sum(g == a for g, a in zip(guesses, fruits.label))
print(f"My rule got {correct} out of 8 right")
```

## Step 3: the computer learns the rule
`model.fit(X, y)` is the moment of learning.

```
from sklearn.tree import DecisionTreeClassifier

X = fruits[["weight", "smoothness"]]   # what we measure
y = fruits["label"]                    # the answers

model = DecisionTreeClassifier()
model.fit(X, y)                        # <-- learning!

correct = (model.predict(X) == y).sum()
print(f"The model got {correct} out of 8 right")
```

## Step 4: what rule did it learn?

```
from sklearn.tree import export_text
print(export_text(model, feature_names=["weight", "smoothness"]))
```

## Step 5: brand new fruits
The model has never seen these.

```
new = pd.DataFrame({"weight": [155, 185, 135], "smoothness": [9, 2, 3]})
new["prediction"] = model.predict(new)
new
```

## Step 6: see it
Above the dashed line: apple. Below: orange. That line is the model.

```
import matplotlib.pyplot as plt

colors = fruits.label.map({"apple": "red", "orange": "orange"})
plt.scatter(fruits.weight, fruits.smoothness, c=colors, s=150)
plt.scatter(new.weight, new.smoothness,
            marker="*", s=400, c="black")
plt.axhline(5.5, ls="--", c="gray")
plt.xlabel("weight (g)")
plt.ylabel("smoothness")
plt.show()
```
